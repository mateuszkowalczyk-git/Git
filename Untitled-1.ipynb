```python
def czy_kolejne(a: int, b: int, c: int) -> bool:
    liczby = sorted([a, b, c])
    if liczby[1] - liczby[0] == 1 and liczby[2] - liczby [1] == 1:
        return True
    else:
        return False

def zamiana_binarna_na_dziesietna(binarna: str) -> int:
    binarna = binarna.strip()
    
    if not all(char in '01' for char in binarna) or len(binarna) == 0:
        raise ValueError(f"'{binarna}' nie jest poprawną liczbą binarną.")

    dziesietna = 0
    for i in range(len(binarna)):
        dziesietna += int(binarna[-(i + 1)]) * (2 ** i)
    return dziesietna

def main():
    a = input("podaj pierwsza liczbe: ")
    b = input("podaj druga liczbe: ")
    c = input("podaj trzecia liczbe: ")

    a = zamiana_binarna_na_dziesietna(a)
    b = zamiana_binarna_na_dziesietna(b)
    c = zamiana_binarna_na_dziesietna(c)

    if czy_kolejne(a,b,c):
        print("liczby sa kolejne")
    else:
        print("liczby nie sa kolejne")
main()
```
